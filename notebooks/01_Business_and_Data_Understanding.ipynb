# Customer Churn Intelligence Platform

## Business Problem

Customer churn occurs when customers stop using a company's products or services. Losing customers directly affects revenue and increases customer acquisition costs.

Instead of reacting after customers leave, companies aim to identify customers who are likely to churn so that retention strategies can be applied proactively.

## Project Objective

Develop a machine learning model that predicts whether a customer is likely to churn and provides business insights to help reduce customer attrition.

## Machine Learning Problem

- Problem Type: Supervised Learning
- Task: Binary Classification
- Target Variable: Churn (Yes/No)

## Success Criteria

- High recall for churned customers
- Good overall model performance
- Explainable predictions
- Actionable business recommendations
