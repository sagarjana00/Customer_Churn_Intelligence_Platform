```python
import joblib
import json
from pathlib import Path

import matplotlib.pyplot as plt

from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score,precision_score,recall_score, f1_score,roc_auc_score,classification_report, ConfusionMatrixDisplay, RocCurveDisplay
```

```python
processed_dir = Path("../data/processed")
x_train = joblib.load(processed_dir/"x_train_processed.pkl")
x_test = joblib.load(processed_dir/"x_test_processed.pkl")
y_train = joblib.load(processed_dir/"y_train.pkl")
y_test = joblib.load(processed_dir/"y_test.pkl")
```

```python
dt = DecisionTreeClassifier(random_state=1)

dt.fit(x_train,y_train)
```

```python
y_pred = dt.predict(x_test)
y_prob = dt.predict_proba(x_test)[:, 1]
```

```python
accuracy = accuracy_score(y_test, y_pred)
accuracy
```

```python
precision = precision_score(y_test, y_pred)
precision
```

```python
recall = recall_score(y_test, y_pred)
recall
```

```python
f1 = f1_score(y_test, y_pred)
f1
```

```python
roc_auc = roc_auc_score(y_test, y_prob)
roc_auc
```

```python
dt_metrics = {
    "Model": "Decision Tree",
    "Accuracy": accuracy,
    "Precision": precision,
    "Recall": recall,
    "F1 Score": f1,
    "ROC-AUC": roc_auc
}

dt_metrics
```

```python
print(classification_report(y_test, y_pred))
```

```python
ConfusionMatrixDisplay.from_predictions(
    y_test,
    y_pred,
    cmap="Blues"
)

plt.title("Decision Tree - Confusion Matrix")
plt.show()
```

```python
RocCurveDisplay.from_predictions(y_test,y_prob)
plt.title("Decision Tree - ROC Curve")
plt.show()
```

```python
models_dir = Path("../models")
joblib.dump(dt, models_dir / "decision_tree.pkl")
```

```python
results_dir = Path("../results")
with open(results_dir / "decision_tree_metrics.json", "w") as f:
    json.dump(dt_metrics, f, indent=4)
```
